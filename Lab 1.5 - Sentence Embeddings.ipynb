```python
from sentence_transformers import SentenceTransformer
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # for 3D plotting

# Load sentence transformer model
sbert = SentenceTransformer("all-MiniLM-L6-v2")

# Sentences for embedding
sentences = [
    # Negative
    "The movie was bad.",
    "The movie was awful.",
    "The food was terrible.",
    "The service was disappointing.",
    
    # Positive
    "The movie was great.",
    "The movie was excellent.",
    "The food was delicious.",
    "The service was amazing.",
    
    # Neutral / Mixed
    "The movie was okay.",
    "The food was average.",
    "The service was fine.",
    "The weather was moderate."
]

# Generate embeddings
embeddings = sbert.encode(sentences)

# Reduce dimensions to 3D using PCA
pca = PCA(n_components=3)
reduced = pca.fit_transform(embeddings)

# 3D plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

for i, sentence in enumerate(sentences):
    x, y, z = reduced[i]
    ax.scatter(x, y, z, label=sentence, s=60)
    ax.text(x, y, z, sentence, fontsize=9)

ax.set_title("Sentence Embeddings (SBERT) - 3D PCA projection")
```
